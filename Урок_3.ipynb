# Пропущенные значения

```python
# Идентификация пропусков осуществляется с помощью метода .isna()
df.MPG.isna()
```

```python
df.MPG.isna().sum() # подсчет количества True, то есть количества пропусков
```

```python
df.isna().sum()
```

```python
df.Horsepower = df.Horsepower.fillna(df.Horsepower.mean()) # заполнение пропущенных значений
#в качестве аргумента указывается значение, которым заполняются пропуски
```
