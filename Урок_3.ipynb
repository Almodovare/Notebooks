```python
df.shape # возвраращает форму набора данных (кол-во строчек;кол-во столбцов)
```

```python
df.info() # возвращает метаинформацию о наборе данных
```

```python
df.describe().round(1) #метод .describe() выводит описательные статистики все сразу
# метод .round() округляет до указанного количества знаков
```

```python
df.mean() # возвращает среднее арифметическое значение
```

```python
df.MPG.mean().round(2) # можно вывести среднее для отдельного столбца
```

```python
df.Displacement.mean().round(3)
```

```python
df.std() # расчет среднего квадратического отклонения
```

```python
df.var() # расчет дисперсии
```

```python
df.median() # расчет медианы
```

```python
df.max() # расчет максимального значения
```

```python
df.min() # расчет минимального значения
```

```python
df.count() # расчет количества наблюдений (n)
```

```python
df.quantile(0.25) #расчет квантиля, например, первого квартиля
# в качестве аргумента может быть указано другое число, например,
# для Q3 = 0.75
```

```python
df.sum() # расчет суммы значений переменной
```

```python
df.skew() # коэффициент асимметрии. Чем он ближе к 0, тем более симметричное
# распределение
```

```python
df.kurt() # коэффицикент эксцесса. Чем ближе к 0, тем распределение ближе к нормальному
# по остроте вершины
```
